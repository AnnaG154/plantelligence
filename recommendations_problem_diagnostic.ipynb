```python
import json, pathlib

with open(pathlib.Path("../artifacts/feature_list.json"), "r", encoding="latin-1") as f:
    fl = json.load(f)

feat_order = fl["features_order"]

cat_cols = [x["name"] for x in feat_order if x["type"] == "categorical"]
num_cols_all = [x["name"] for x in feat_order if x["type"] == "numeric"]

# Zgodnie z Twoją wcześniejszą notatką: Scaler był trenowany na 2 kolumnach,
# a 'toxicity_any' (binarka) była dodawana "na końcu", bez skalowania.
num_scaled = [c for c in num_cols_all if c != "toxicity_any"]
bin_cols = ["toxicity_any"] if "toxicity_any" in num_cols_all else []

print("Kategoryczne (OHE):", cat_cols)
print("Numeryczne skalowane (2):", num_scaled)
print("Binarka (bez skalowania):", bin_cols)
```
