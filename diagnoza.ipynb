```python
import json
from pathlib import Path
import pandas as pd

# ścieżki
DATA_PATH = Path(r"c:/Users/annas/projekty_infoshare/projekt ML/data/plants_clean_final.csv")
ARTIFACTS_JSON = Path(r"c:/Users/annas/projekty_infoshare/projekt ML/artifacts/feature_list.json")

# 1) walidacja kolumn
df = pd.read_csv(DATA_PATH)
with open(ARTIFACTS_JSON, "r", encoding="latin-1") as f:
    fl = json.load(f)

expected = [f["name"] for f in fl["features_order"]]
missing = [c for c in expected if c not in df.columns]
print("Brakujące kolumny:", missing if missing else "brak — jest OK ✅")

# 2) szybki podgląd JSON (pierwsze 300 znaków)
content = ARTIFACTS_JSON.read_text(encoding="latin-1")
print("Podgląd JSON:\n", content[:300], "...")
```
